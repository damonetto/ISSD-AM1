# Dataset de analisis de sentimientos

```python
import requests
import zipfile
import io
import pandas as pd

url_sentiment = "https://archive.ics.uci.edu/static/public/331/sentiment+labelled+sentences.zip"

# Download the zip file
print(f"Descargando {url_sentiment}...")
response = requests.get(url_sentiment)
response.raise_for_status() # Raise an exception for HTTP errors

# Extract the contents of the zip file
with zipfile.ZipFile(io.BytesIO(response.content)) as zf:
    zf.extractall('.') # Extract to the current directory

print('Dataset de sentimientos descargado y extraído correctamente.')

# Load one of the text files as a DataFrame
# The zip usually contains imdb_labelled.txt, amazon_cells_labelled.txt, yelp_labelled.txt
# These are tab-separated and typically have no header.
csv_file_name = 'sentiment labelled sentences/imdb_labelled.txt'

# Load the file into a DataFrame named 'dataset'
dataset = pd.read_csv(csv_file_name, sep='\t', header=None, names=['text', 'sentiment'])

print(f"Cargado '{csv_file_name}' en la variable 'dataset'.")
print(dataset.head())
```

```python
csv_file_name_amazon = 'sentiment labelled sentences/amazon_cells_labelled.txt'
dataset_amazon = pd.read_csv(csv_file_name_amazon, sep='\t', header=None, names=['text', 'sentiment'])
print(f"Cargado '{csv_file_name_amazon}' en la variable 'dataset_amazon'.")
print(dataset_amazon.head())
```

```python
csv_file_name_yelp = 'sentiment labelled sentences/yelp_labelled.txt'
dataset_yelp = pd.read_csv(csv_file_name_yelp, sep='\t', header=None, names=['text', 'sentiment'])
print(f"Cargado '{csv_file_name_yelp}' en la variable 'dataset_yelp'.")
print(dataset_yelp.head())
```
